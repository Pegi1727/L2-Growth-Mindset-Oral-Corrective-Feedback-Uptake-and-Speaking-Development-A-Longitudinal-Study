# روند نمره گفتاری

مقایسه نمره گفتاری در سه نقطه زمانی و محاسبه تغییرات.

**Data source:** `Longitudinal_Data_120.csv` (120 participants).

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('Longitudinal_Data_120.csv')
time_cols = ['T1_Speaking','T2_Speaking','T3_Speaking']
```

```python
display(df[time_cols].agg(['mean','std','median','min','max']).T)
```

```python
means = df[time_cols].mean()
ax = means.plot(marker='o', figsize=(7,4), title='Mean Speaking Score Over Time')
ax.set_ylabel('Mean score'); ax.set_xlabel('Time point'); plt.grid(alpha=.25); plt.show()
```

```python
df['Change_T1_T3'] = df['T3_Speaking'] - df['T1_Speaking']
print(df['Change_T1_T3'].describe())
print('Improved:', (df['Change_T1_T3']>0).sum(), 'Unchanged:', (df['Change_T1_T3']==0).sum(), 'Declined:', (df['Change_T1_T3']<0).sum())
```
