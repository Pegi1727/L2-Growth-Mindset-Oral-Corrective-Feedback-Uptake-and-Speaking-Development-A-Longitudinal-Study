# ذهنیت و پذیرش بازخورد

توصیف ذهنیت و نرخ پذیرش و بررسی ارتباط آن‌ها با تغییر گفتاری.

**Data source:** `Longitudinal_Data_120.csv` (120 participants).

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('Longitudinal_Data_120.csv')
df['Speaking_Change'] = df['T3_Speaking'] - df['T1_Speaking']
```

```python
display(df[['Mindset','Uptake_Rate','Speaking_Change']].describe().T)
```

```python
print('Correlation matrix:')
display(df[['Mindset','Uptake_Rate','Speaking_Change']].corr().round(3))
```

```python
ax = df.plot.scatter(x='Mindset', y='Uptake_Rate', alpha=.65, figsize=(6,4), title='Mindset vs Feedback Uptake')
plt.show()
```
