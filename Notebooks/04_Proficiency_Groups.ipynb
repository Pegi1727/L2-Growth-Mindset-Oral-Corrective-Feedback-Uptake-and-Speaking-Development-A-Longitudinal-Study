# مقایسه سطوح مهارت

تقسیم شرکت‌کنندگان به گروه‌های مهارتی و مقایسه میانگین نمرات.

**Data source:** `Longitudinal_Data_120.csv` (120 participants).

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('Longitudinal_Data_120.csv')
df['Proficiency_Group'] = pd.qcut(df['Proficiency'], q=3, labels=['Low','Medium','High'], duplicates='drop')
time_cols=['T1_Speaking','T2_Speaking','T3_Speaking']
```

```python
display(df.groupby('Proficiency_Group', observed=False)[['Proficiency']+time_cols].mean().round(2))
```

```python
group_means=df.groupby('Proficiency_Group', observed=False)[time_cols].mean().T
ax=group_means.plot(marker='o', figsize=(8,4), title='Speaking Scores by Proficiency Group')
ax.set_ylabel('Mean speaking score'); plt.grid(alpha=.25); plt.show()
```
