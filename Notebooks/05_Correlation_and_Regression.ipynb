# ارتباط‌ها و رگرسیون

محاسبه همبستگی‌ها و برازش یک مدل خطی ساده برای تغییر نمره.

**Data source:** `Longitudinal_Data_120.csv` (120 participants).

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('Longitudinal_Data_120.csv')
df['Speaking_Change'] = df['T3_Speaking'] - df['T1_Speaking']
predictors=['Proficiency','Mindset','Uptake_Rate']
```

```python
display(df[predictors+['Speaking_Change']].corr().round(3))
```

```python
# مدل OLS در صورت نصب statsmodels
try:
    import statsmodels.api as sm
    X=sm.add_constant(df[predictors])
    model=sm.OLS(df['Speaking_Change'], X).fit()
    print(model.summary())
except ImportError:
    print('Install statsmodels to run OLS: pip install statsmodels')
```

```python
ax=df.plot.scatter(x='Uptake_Rate', y='Speaking_Change', alpha=.65, figsize=(6,4), title='Uptake Rate vs Speaking Change')
plt.show()
```
